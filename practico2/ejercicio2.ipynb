#### 2.b)

show_id: Identificador único para cada película o serie en el catálogo.

Tipo de dato: Cualitativo (categórico) nominal.

type: Indica si la producción es una Película ("Movie") o un Programa de TV/Serie ("TV Show").

Tipo de dato: Cualitativo (categórico) nominal.

title: El título de la película o serie.

Tipo de dato: Cualitativo (categórico) nominal (texto libre).

director: Nombre del director o directores de la obra.

Tipo de dato: Cualitativo (categórico) nominal.

cast: Lista de actores y actrices que participan.

Tipo de dato: Cualitativo (categórico) nominal.

country: País o países donde se produjo la película o serie.

Tipo de dato: Cualitativo (categórico) nominal.

date_added: La fecha exacta en la que se agregó la producción a la plataforma de Netflix.

Tipo de dato: Cualitativo ordinal (o cuantitativo si se lo procesa como formato "fecha").

release_year: El año en el que se lanzó originalmente la producción.

Tipo de dato: Cuantitativo numérico discreto.

rating: La clasificación por edades de la producción (ej. TV-MA, PG-13).

Tipo de dato: Cualitativo (categórico) ordinal (ya que implica una jerarquía de edad permitida).

duration: La duración de la producción (en minutos para las películas y en cantidad de temporadas para las series).

Tipo de dato: Cuantitativo discreto (cantidad de minutos/temporadas).

listed_in: Las categorías o géneros en los que Netflix clasifica la producción (ej. Documentales, Comedias).

Tipo de dato: Cualitativo (categórico) nominal.

description: Una breve sinopsis o resumen de la trama.

Tipo de dato: Cualitativo (categórico) nominal (datos no estructurados de texto libre).

```python
# i) Cuántos directores distintos hay
num_directores = netflix["director"].nunique()
print(f"Hay {num_directores} directores distintos")
print(" ")
# ojo: si un título tiene varios directores juntos ("Fulano, Mengano"),
# Pandas lo cuenta como una combinación distinta, no como directores individuales

# ii) Director con más películas
conteo_directores = netflix["director"].value_counts()
print("El director con mas peliculas es: ", conteo_directores.idxmax()," con ", conteo_directores.max(), " peliculas.")
print(" ")

# iii) Cuántas películas y cuántos programas de TV hay
print(netflix["type"].value_counts())
print(" ")

# iv) Película y serie más vieja
print("Pelicula mas vieja:")
peliculas = netflix[netflix["type"] == "Movie"]
mas_vieja_pelicula = peliculas.loc[peliculas["release_year"].idxmin()]
print(mas_vieja_pelicula[["title", "release_year"]])
print(" ")
print("Serie mas vieja:")
series = netflix[netflix["type"] == "TV Show"]
mas_vieja_serie = series.loc[series["release_year"].idxmin()]
print(mas_vieja_serie[["title", "release_year"]])
print(" ")

# v) Cantidad de películas de un país dado
# Armamos la lista de países únicos (separando los que vienen combinados)
paises_individuales = (
    netflix["country"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
    .unique()
)

paises_individuales = sorted(paises_individuales)

# Mostramos la lista de países disponibles
print("Países disponibles en el catálogo:")
for pais in paises_individuales:
    print(f"- {pais}")

# Le pedimos al usuario que ingrese un país
pais_elegido = input("\nIngresá el país que querés consultar: ")

# Validamos que el país ingresado exista en la lista
if pais_elegido not in paises_individuales:
    print(f"'{pais_elegido}' no está en la lista de países del catálogo.")
else:
    # Contamos las películas de ese país
    filtro = netflix["country"].str.contains(pais_elegido, case=False, na=False)
    cantidad = netflix[filtro & (netflix["type"] == "Movie")].shape[0]
    print(f"\nCantidad de películas de {pais_elegido}: {cantidad}")
```
