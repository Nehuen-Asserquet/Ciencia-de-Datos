```python
# i) Cuántos directores distintos hay
num_directores = netflix["director"].nunique()
print(f"Hay {num_directores} directores distintos")
print(" ")
# ojo: si un título tiene varios directores juntos ("Fulano, Mengano"),
# Pandas lo cuenta como una combinación distinta, no como directores individuales

# ii) Director con más películas
conteo_directores = netflix["director"].value_counts()
print("El director con mas peliculas es: ", conteo_directores.idxmax()," con ", conteo_directores.max(), " peliculas.")
print(" ")

# iii) Cuántas películas y cuántos programas de TV hay
print(netflix["type"].value_counts())
print(" ")

# iv) Película y serie más vieja
print("Pelicula mas vieja:")
peliculas = netflix[netflix["type"] == "Movie"]
mas_vieja_pelicula = peliculas.loc[peliculas["release_year"].idxmin()]
print(mas_vieja_pelicula[["title", "release_year"]])
print(" ")
print("Serie mas vieja:")
series = netflix[netflix["type"] == "TV Show"]
mas_vieja_serie = series.loc[series["release_year"].idxmin()]
print(mas_vieja_serie[["title", "release_year"]])
print(" ")

# v) Cantidad de películas de un país dado
# Armamos la lista de países únicos (separando los que vienen combinados)
paises_individuales = (
    netflix["country"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
    .unique()
)

paises_individuales = sorted(paises_individuales)

# Mostramos la lista de países disponibles
print("Países disponibles en el catálogo:")
for pais in paises_individuales:
    print(f"- {pais}")

# Le pedimos al usuario que ingrese un país
pais_elegido = input("\nIngresá el país que querés consultar: ")

# Validamos que el país ingresado exista en la lista
if pais_elegido not in paises_individuales:
    print(f"'{pais_elegido}' no está en la lista de países del catálogo.")
else:
    # Contamos las películas de ese país
    filtro = netflix["country"].str.contains(pais_elegido, case=False, na=False)
    cantidad = netflix[filtro & (netflix["type"] == "Movie")].shape[0]
    print(f"\nCantidad de películas de {pais_elegido}: {cantidad}")
```

##### 2.b) TP 3
Hipótesis 1: Las películas (Movie) lanzadas en años más recientes (release_year) son más cortas en duración (duration_min).

Hipótesis 2: Existe una relación positiva entre el año de lanzamiento de un contenido (release_year) y el año en que fue incorporado al catálogo de Netflix (year_added).

Hipótesis 3: Las películas más largas (duration_min) tienen sinopsis o descripciones más extensas (description_length).

#### 2.c) TP 3

```python
# --- Hipótesis 1: release_year vs duration_min (en Películas) ---
corr1 = peliculas['release_year'].corr(peliculas['duration_min'])
print(f"Correlación 1 (release_year vs duration_min): {corr1:.4f}")

plt.figure(figsize=(6, 4))
sns.scatterplot(data=peliculas, x='release_year', y='duration_min')
plt.title(f"Relacion entre Año de Lanzamiento y Duración en Minutos (Correlacion = {corr1:.2f})")
plt.xlabel("Año de Lanzamiento (release_year)")
plt.ylabel("Duración en Minutos (duration_min)")
plt.show()


# 1. Definir df_netflix y extraer el año desde la columna date_added
df_netflix = netflix.copy()
df_netflix['year_added'] = pd.to_datetime(df_netflix['date_added'], errors='coerce').dt.year

# 2. Hipótesis 2: release_year vs year_added
corr2 = df_netflix['release_year'].corr(df_netflix['year_added'])
print(f"Correlación 2 (release_year vs year_added): {corr2:.4f}")

plt.figure(figsize=(6, 4))
sns.scatterplot(data=df_netflix, x='release_year', y='year_added')
plt.title(f"Relacion entre Año de Lanzamiento y Año de Agregado (Correlacion = {corr2:.2f})")
plt.xlabel("Año de Lanzamiento (release_year)")
plt.ylabel("Año de Agregado a Netflix (year_added)")
plt.show()


# --- Hipótesis 3: duration_min vs description_length ---
corr3 = peliculas['duration_min'].corr(peliculas['description_length'])
print(f"Correlación 3 (duration_min vs description_length): {corr3:.4f}")

plt.figure(figsize=(6, 4))
sns.scatterplot(data=peliculas, x='duration_min', y='description_length')
plt.title(f"Relacion entre Duración de Película y Largo de Descripción (Correlacion = {corr3:.2f})")
plt.xlabel("Duración en Minutos (duration_min)")
plt.ylabel("Cantidad de Caracteres en Descripción")
plt.show()
```

#### justificacion:
Hipótesis 1 ($release\_year$ vs $duration\_min$): La correlación es débil o ligeramente negativa (alrededor de $-0.20$). El gráfico muestra que, si bien hay mucha variabilidad, las películas muy antiguas solían tener duraciones más dispersas y las recientes tienden a concentrarse cerca de los 90–100 minutos. La hipótesis es parcialmente verosímil.  

Hipótesis 2 ($release\_year$ vs $year\_added$): Muestra una correlación positiva moderada (alrededor de $0.50$). El scatter plot confirma una clara tendencia diagonal: la mayoría del contenido agregado recientemente también fue producido en años recientes. La hipótesis es muy verosímil.  

Hipótesis 3 ($duration\_min$ vs $description\_length$): La correlación es prácticamente cero (alrededor de $0.01$). En la gráfica de dispersión la nube de puntos es totalmente amorfa, lo que indica que la duración de una película no influye en la longitud del texto descriptivo. La hipótesis no es verosímil.
